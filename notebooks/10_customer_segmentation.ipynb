# Customer Segmentation (Module 10)

## Objective
Apply K-Means clustering to segment enterprise customer profiles based on annual revenue, total revenue, average order value, and engagement activity scores. Evaluate performance via Silhouette score and Davies-Bouldin metrics, and reduce dimension for visual review using PCA.

```python
from pathlib import Path
import sys

import pandas as pd

current_dir = Path.cwd().resolve()
project_root = current_dir.parent if current_dir.name == 'notebooks' else current_dir
backend_root = project_root / 'backend'
sys.path.insert(0, str(backend_root))

from app.ml.customer_segmentation import CustomerSegmentationPaths, CustomerSegmentationService

features_dir = project_root / 'features'
trained_models_dir = project_root / 'trained_models'
reports_dir = project_root / 'reports'
exports_dir = project_root / 'exports'

paths = CustomerSegmentationPaths(
    features_dir=features_dir,
    trained_models_dir=trained_models_dir,
    reports_dir=reports_dir,
    exports_dir=exports_dir
)
service = CustomerSegmentationService(paths)
service
```

## Segment Customers

We will run KMeans and generate clusters.

```python
artifact, segments_df = service.train_and_register(n_clusters=4)
print(f"Silhouette Score: {artifact.silhouette:.4f}")
print(f"Davies-Bouldin Index: {artifact.davies_bouldin:.4f}")
```
