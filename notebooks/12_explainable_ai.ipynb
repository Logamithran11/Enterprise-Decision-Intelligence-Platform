# Explainable AI with SHAP (Module 12)

## Objective
Implement SHAP tree explainers on registered tree models. Output global summary plots (feature importance and density impact) and local waterfall/force plots showing individual feature contributions to a specific prediction.

```python
from pathlib import Path
import sys

import pandas as pd

current_dir = Path.cwd().resolve()
project_root = current_dir.parent if current_dir.name == 'notebooks' else current_dir
backend_root = project_root / 'backend'
sys.path.insert(0, str(backend_root))

from app.ml.explainability import ExplainabilityService

trained_models_dir = project_root / 'trained_models'
exports_dir = project_root / 'exports'
features_dir = project_root / 'features'

service = ExplainabilityService(trained_models_dir=trained_models_dir, exports_dir=exports_dir)
service
```

## Global Explanations

We will load customer features and compute global SHAP importances for the churn prediction model.

```python
customer_features = pd.read_csv(features_dir / 'customer_features.csv')
global_exps = service.explain_global('churn_prediction_model', customer_features.head(100))
pd.DataFrame(global_exps['global_importance']).head(10)
```

## Local Explanations

Let's explain a single customer prediction to see why they were predicted to churn (or not).

```python
sample_customer = customer_features.iloc[[0]]
local_exps = service.explain_local('churn_prediction_model', sample_customer)
print("Base value prediction:", local_exps['base_value'])
pd.DataFrame(local_exps['features']).head(10)
```
